```python


import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms
from torchvision.utils import make_grid
from torch.utils.data import DataLoader
import matplotlib.pyplot as plt
import numpy as np

# ─── Hyperparameters ──────────────────────────────────────────────────────────
latent_dim   = 100
epochs       = 100
batch_size   = 128
lr           = 0.0002
beta1        = 0.5
image_size   = 28
channels     = 1
device       = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")

# ─── Data ─────────────────────────────────────────────────────────────────────
transform = transforms.Compose([
    transforms.Resize(image_size),
    transforms.ToTensor(),
    transforms.Normalize([0.5], [0.5]),
])

train_dataset = datasets.MNIST(root="./data", train=True,
                                transform=transform, download=True)
train_loader  = DataLoader(train_dataset, batch_size=batch_size, shuffle=True,
                           num_workers=2, pin_memory=True)


# ─── Task 2: Convolutional Generator (DCGAN-style)
class Generator(nn.Module):
    """
    Replaces the original fully-connected generator with transposed
    convolutions (DC-GAN architecture).

    Latent vector (100,) → reshape → conv layers → 1×28×28 image
    """
    def __init__(self, latent_dim: int):
        super().__init__()
        # Project and reshape
        self.project = nn.Sequential(
            nn.Linear(latent_dim, 256 * 7 * 7),
            nn.BatchNorm1d(256 * 7 * 7),
            nn.ReLU(True),
        )
        # Transposed-conv upsampling: 7×7 → 14×14 → 28×28
        self.conv = nn.Sequential(
            # Block 1: 256ch 7×7 → 128ch 14×14
            nn.ConvTranspose2d(256, 128, kernel_size=4, stride=2, padding=1, bias=False),
            nn.BatchNorm2d(128),
            nn.ReLU(True),
            # Block 2: 128ch 14×14 → 64ch 28×28
            nn.ConvTranspose2d(128, 64, kernel_size=4, stride=2, padding=1, bias=False),
            nn.BatchNorm2d(64),
            nn.ReLU(True),
            # Task 1 – extra conv layer for richer features
            nn.Conv2d(64, 64, kernel_size=3, stride=1, padding=1, bias=False),
            nn.BatchNorm2d(64),
            nn.ReLU(True),
            # Output: 1ch 28×28, values in [-1, 1]
            nn.Conv2d(64, channels, kernel_size=3, stride=1, padding=1),
            nn.Tanh(),
        )

    def forward(self, z: torch.Tensor) -> torch.Tensor:
        x = self.project(z)
        x = x.view(x.size(0), 256, 7, 7)
        return self.conv(x)


# ─── Task 2: Convolutional Discriminator (DCGAN-style) ───────────────────────
class Discriminator(nn.Module):
    """
    Replaces the original fully-connected discriminator with strided
    convolutions (DC-GAN architecture).

    1×28×28 image → conv layers → scalar probability
    """
    def __init__(self):
        super().__init__()
        self.conv = nn.Sequential(
            # Block 1: 1ch 28×28 → 64ch 14×14
            nn.Conv2d(channels, 64, kernel_size=4, stride=2, padding=1, bias=False),
            nn.LeakyReLU(0.2, inplace=True),
            # Block 2: 64ch 14×14 → 128ch 7×7
            nn.Conv2d(64, 128, kernel_size=4, stride=2, padding=1, bias=False),
            nn.BatchNorm2d(128),
            nn.LeakyReLU(0.2, inplace=True),
            # Task 1 – extra conv layer
            nn.Conv2d(128, 256, kernel_size=3, stride=1, padding=1, bias=False),
            nn.BatchNorm2d(256),
            nn.LeakyReLU(0.2, inplace=True),
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(256 * 7 * 7, 1),
            nn.Sigmoid(),
        )

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.classifier(self.conv(x))


# ─── Weight initialisation (DCGAN paper recommendation) ──────────────────────
def weights_init(m):
    classname = m.__class__.__name__
    if "Conv" in classname:
        nn.init.normal_(m.weight.data, 0.0, 0.02)
    elif "BatchNorm" in classname:
        nn.init.normal_(m.weight.data, 1.0, 0.02)
        nn.init.constant_(m.bias.data, 0)


generator     = Generator(latent_dim).to(device)
discriminator = Discriminator().to(device)
generator.apply(weights_init)
discriminator.apply(weights_init)

# ─── Loss & Optimisers ────────────────────────────────────────────────────────
criterion             = nn.BCELoss()
optimizer_G           = optim.Adam(generator.parameters(),     lr=lr, betas=(beta1, 0.999))
optimizer_D           = optim.Adam(discriminator.parameters(), lr=lr, betas=(beta1, 0.999))

# Fixed noise for consistent visualisation across epochs
fixed_noise = torch.randn(16, latent_dim, device=device)


# ─── Visualisation helper ────────────────────────────────────────────────────
def visualize_images(gen, noise, epoch: int, save: bool = False):
    gen.eval()
    with torch.no_grad():
        imgs = gen(noise).cpu()                        # (16, 1, 28, 28)
    grid = make_grid(imgs, nrow=4, normalize=True).permute(1, 2, 0).numpy()
    plt.figure(figsize=(6, 6))
    plt.imshow(grid, cmap="gray")
    plt.axis("off")
    plt.title(f"Epoch {epoch}")
    if save:
        plt.savefig(f"generated_epoch_{epoch:03d}.png", bbox_inches="tight")
    plt.show()
    gen.train()


# ─── Training loop ───────────────────────────────────────────────────────────
g_losses, d_losses = [], []

def train_gan():
    for epoch in range(epochs):
        epoch_d, epoch_g = 0.0, 0.0
        for real_images, _ in train_loader:
            real_images = real_images.to(device)
            current_bs  = real_images.size(0)

            # ── Label smoothing (improves GAN stability) ──────────────────
            real_labels = torch.full((current_bs, 1), 0.9, device=device)   # soft real
            fake_labels = torch.zeros(current_bs, 1, device=device)

            # ── Train Discriminator ───────────────────────────────────────
            optimizer_D.zero_grad()

            real_loss = criterion(discriminator(real_images), real_labels)

            noise      = torch.randn(current_bs, latent_dim, device=device)
            fake_imgs  = generator(noise).detach()
            fake_loss  = criterion(discriminator(fake_imgs), fake_labels)

            d_loss = real_loss + fake_loss
            d_loss.backward()
            optimizer_D.step()

            # ── Train Generator ───────────────────────────────────────────
            optimizer_G.zero_grad()

            noise     = torch.randn(current_bs, latent_dim, device=device)
            fake_imgs = generator(noise)
            # Generator wants discriminator to output 1 (real) for its fakes
            g_loss = criterion(discriminator(fake_imgs),
                               torch.ones(current_bs, 1, device=device))
            g_loss.backward()
            optimizer_G.step()

            epoch_d += d_loss.item()
            epoch_g += g_loss.item()

        avg_d = epoch_d / len(train_loader)
        avg_g = epoch_g / len(train_loader)
        d_losses.append(avg_d)
        g_losses.append(avg_g)

        print(f"Epoch [{epoch+1:>3}/{epochs}] | D Loss: {avg_d:.4f} | G Loss: {avg_g:.4f}")

        # Visualise every 10 epochs
        if (epoch + 1) % 10 == 0:
            visualize_images(generator, fixed_noise, epoch + 1, save=True)

    # ── Plot training losses ─────────────────────────────────────────────
    plt.figure(figsize=(8, 4))
    plt.plot(d_losses, label="Discriminator Loss")
    plt.plot(g_losses, label="Generator Loss")
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.title("GAN Training Losses")
    plt.legend()
    plt.tight_layout()
    plt.savefig("training_losses.png")
    plt.show()


# ─── Generate final images ───────────────────────────────────────────────────
def generate_images(gen, n: int = 16):
    gen.eval()
    noise = torch.randn(n, latent_dim, device=device)
    with torch.no_grad():
        imgs = gen(noise).cpu()
    grid = make_grid(imgs, nrow=4, normalize=True).permute(1, 2, 0).numpy()
    plt.figure(figsize=(6, 6))
    plt.imshow(grid, cmap="gray")
    plt.axis("off")
    plt.title("Final Generated Images")
    plt.savefig("final_generated.png", bbox_inches="tight")
    plt.show()


# ─── Entry point ─────────────────────────────────────────────────────────────
if __name__ == "__main__":
    print("=== Model Summaries ===")
    total_g = sum(p.numel() for p in generator.parameters())
    total_d = sum(p.numel() for p in discriminator.parameters())
    print(f"Generator params     : {total_g:,}")
    print(f"Discriminator params : {total_d:,}")
    print()

    train_gan()
    generate_images(generator)
```
