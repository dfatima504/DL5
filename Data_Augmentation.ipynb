```python
import torch
from torchvision import transforms
from PIL import Image
import os
import matplotlib.pyplot as plt

save_folder = "PyTorch_Augmented_Images"
os.makedirs(save_folder, exist_ok=True)


augmentation_pipeline = transforms.Compose([
    transforms.RandomRotation(degrees=40),
    transforms.RandomAffine(degrees=0, shear=20),
    transforms.RandomResizedCrop(size=(224, 224), scale=(0.8, 1.2)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.ColorJitter(brightness=(0.5, 1.5))
])

image_path = "images.jpg"
try:
    img = Image.open(image_path).convert("RGB")

    # Step 4: Generate and Save 40 Augmented Images
    print(f"Generating 40 images in {save_folder}...")
    for i in range(40):
        # Apply the random transformations
        augmented_img = augmentation_pipeline(img)

        # Save the image to the directory
        file_name = f"aug_image_{i+1}.jpg"
        save_path = os.path.join(save_folder, file_name)
        augmented_img.save(save_path)

        # Optional: Visualize the first 3 images
        if i < 3:
            plt.figure()
            plt.imshow(augmented_img)
            plt.title(f"Augmented Version {i+1}")
            plt.axis('off')
            plt.show()

    print(f"Task Complete. Images saved at: {os.path.abspath(save_folder)}")

except FileNotFoundError:
    print("Error: Please ensure 'images.jpg' exists in the script directory.")
```
